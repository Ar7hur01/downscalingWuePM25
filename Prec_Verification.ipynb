In this Notebook the Precipitation will be verified due to 0.0 % Feature IMportance which the RF-Results show?
This is most probably due to clouds (when it rains) and therefore a lack of Satellite AOD and NO2 data.

```python
import pandas as pd

# 1. Datei einlesen
file_path = r"C:\Users\AD\Desktop\Uni Würzburg\EAGLE\InnoLab\new_Output\IDW_hoch2\Exp2_Features_RH_PM25_NDVI_TPI_S5P_ERA5_MAIAC_Verkehr_Wetter.csv"
df = pd.read_csv(file_path, sep=";", decimal=",")

# 2. Alle Zeilen filtern, an denen es geregnet hat (Niederschlag > 0)
df_rain = df[df["icon_precip_12h"] > 0]
total_rain_days = len(df_rain)

# 3. Zählen, wie oft an diesen Regentagen Satellitendaten vorhanden sind.
# Wir prüfen, ob die Werte NICHT leer (notna) und NICHT 0 sind.
valid_sat_during_rain = df_rain[
    (df_rain["MAIAC_AOD"].notna()) & (df_rain["MAIAC_AOD"] != 0) &
    (df_rain["S5P_NO2"].notna()) & (df_rain["S5P_NO2"] != 0)
]
count_valid = len(valid_sat_during_rain)

# 4. Ergebnisse ausgeben
print(f"Anzahl der Messungen mit Regen (> 0 mm): {total_rain_days}")
print(f"Davon MIT vorhandenen Satellitendaten (AOD & NO2): {count_valid}")
print(f"Davon OHNE Satellitendaten (Wolken-Blockade): {total_rain_days - count_valid}")
```
